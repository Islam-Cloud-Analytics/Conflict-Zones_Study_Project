```sql
SELECT
  COUNTRY,
  ADMIN1 AS city,
  YEAR(to_date(WEEK, 'dd-MMMM-yyyy')) AS year,
  COUNT(*) AS riot_count
FROM workspace.data_analysis_study_project.`me-conflict-zones-study-project`
WHERE
  EVENT_TYPE = 'Riots'
  AND to_date(WEEK, 'dd-MMMM-yyyy') >= add_months(CURRENT_DATE(), -24)
GROUP BY
  COUNTRY,
  city,
  YEAR(to_date(WEEK, 'dd-MMMM-yyyy'))
ORDER BY
  COUNTRY,
  city,
  year;
```
