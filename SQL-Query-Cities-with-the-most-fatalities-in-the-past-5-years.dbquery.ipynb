```sql
WITH city_fatalities AS (
  SELECT
    COUNTRY,
    ADMIN1 AS CITY,
    any_value(CENTROID_LATITUDE) AS CENTROID_LATITUDE,
    any_value(CENTROID_LONGITUDE) AS CENTROID_LONGITUDE,
    SUM(FATALITIES) AS TOTAL_FATALITIES
  FROM
    workspace.data_analysis_study_project.`me-conflict-zones-study-project`
  WHERE
    to_date(WEEK, 'dd-MMMM-yyyy') >= date_add(current_date(), -5 * 365)
  GROUP BY
    COUNTRY,
    ADMIN1
),
avg_city_fatalities AS (
  SELECT
    AVG(TOTAL_FATALITIES) AS avg_fatalities
  FROM
    city_fatalities
)
SELECT
  COUNTRY,
  CITY,
  CENTROID_LATITUDE,
  CENTROID_LONGITUDE,
  TOTAL_FATALITIES
FROM
  city_fatalities,
  avg_city_fatalities
WHERE
  TOTAL_FATALITIES > avg_city_fatalities.avg_fatalities
ORDER BY
  TOTAL_FATALITIES DESC
```
